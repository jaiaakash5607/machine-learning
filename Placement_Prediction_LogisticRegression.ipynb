```python
import numpy as np
import pandas as pd
import seaborn as sms
```

```python
data=pd.read_csv("/content/placementdata.csv")
data
```

```python
data.info()
```

```python
data.drop(["StudentID"],inplace=True,axis=1)
```

```python
data.PlacementStatus=[1 if x=="Placed" else 0 for x in data.PlacementStatus]
data.ExtracurricularActivities=[1 if x=="Yes" else 0 for x in data.ExtracurricularActivities]
```

```python
data
```

```python
data.PlacementTraining=[1 if x=="Yes" else 0 for x in data.PlacementTraining]
```

```python
data
```

```python
y=data["PlacementStatus"]
x=data.drop(["PlacementStatus"],axis=1)
```

```python
x.hist(figsize=(15,8))
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
x=sc.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
lr.fit(x_train,y_train)
```

```python
lr.score(x_train,y_train)
```

```python
lr.score(x_test,y_test)
```

```python
pre=lr.predict(x_test)
pre[:5]
```
